#Mini Project: Student Management System<br>

Project Title:<br>
Student Management System Using OOP Concepts<br>

Problem Statement:<br>
Develop a Student Management System using Python that demonstrates the concepts of Class, Object, Encapsulation, Inheritance, and Abstraction. The system should allow users to store student information, calculate grades, and display student details based on different student categories.<br>

Requirements<br>

1. Create an Abstract Class Student<br>
The abstract class should contain:<br>
Student ID<br>
Student Name<br>
A private attribute__marks<br>
A method to set marks<br>
A method to get marks<br>
An abstract method calculate_grade()<br>
A method display_details()


```python
from abc import ABC, abstractmethod

class Student(ABC):
    def __init__(self, student_id, student_name):
        self.student_id = student_id
        self.student_name = student_name
        self.__marks = 0

    def set_marks(self, marks):
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    @abstractmethod
    def calculate_grade(self):
        pass

    def display_details(self):
        print("Student ID:", self.student_id)
        print("Student Name:", self.student_name)
        print("Marks:", self.get_marks())


class RegularStudent(Student):
    def calculate_grade(self):
        marks = self.get_marks()

        if marks >= 90:
            return "A+"
        elif marks >= 80:
            return "A"
        elif marks >= 70:
            return "B+"
        elif marks >= 60:
            return "B"
        elif marks >= 50:
            return "C+"
        elif marks >= 40:
            return "C"
        else:
            return "F"

    def display_details(self):
        super().display_details()
        print("Category: Regular Student")
        print("Grade:", self.calculate_grade())
        print()


class ScholarshipStudent(Student):
    def calculate_grade(self):
        marks = self.get_marks()

        if marks >= 85:
            return "A+"
        elif marks >= 75:
            return "A"
        elif marks >= 65:
            return "B+"
        elif marks >= 55:
            return "B"
        elif marks >= 45:
            return "C+"
        elif marks >= 35:
            return "C"
        else:
            return "F"

    def display_details(self):
        super().display_details()
        print("Category: Scholarship Student")
        print("Grade:", self.calculate_grade())
        print()


students = []

while True:
    print("===== Student Management System =====")
    print("1. Add Regular Student")
    print("2. Add Scholarship Student")
    print("3. Display All Students")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        sid = input("Enter Student ID: ")
        name = input("Enter Student Name: ")
        marks = float(input("Enter Marks: "))

        student = RegularStudent(sid, name)
        student.set_marks(marks)
        students.append(student)

        print("Regular Student Added Successfully.\n")

    elif choice == "2":
        sid = input("Enter Student ID: ")
        name = input("Enter Student Name: ")
        marks = float(input("Enter Marks: "))

        student = ScholarshipStudent(sid, name)
        student.set_marks(marks)
        students.append(student)

        print("Scholarship Student Added Successfully.\n")

    elif choice == "3":
        if len(students) == 0:
            print("No student records found.\n")
        else:
            print("\n===== Student Details =====")
            for student in students:
                student.display_details()

    elif choice == "4":
        print("Thank You!")
        break

    else:
        print("Invalid Choice.\n")
```
