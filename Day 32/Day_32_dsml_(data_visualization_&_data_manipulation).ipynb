#Central Tendency

Central tendency is a statistical method used to represent the center or typical value of a dataset. The three main measures are:

Mean: Sum of all observations divided by the number of observations.

Mean=
n
∑x
	​

Median: The middle value when data is arranged in ascending or descending order.
Mode: The value that occurs most frequently in the dataset.

In short: Mean = average, Median = middle value, Mode = most frequent value.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
x = np.random.standard_t(500,size=1000)
```

```python
x
```

```python
x[:15]
```

```python
_mean = np.mean(x)
```

```python
print(f"mean of x : {_mean}")
```

```python
plt.hist(x,bins=20,color="blue")
```

```python
plt.axvline(x=_mean,color="red")
```

```python
plt.hist(x,bins=20,color="blue")
plt.axvline(x=_mean,color="red")
```

#median

```python
_median = np.median(x)
print(f"median of x : {_median}")
```

```python
plt.hist(x,bins=20,color="blue")
```

```python
plt.axvline(x=_median,color="red")
```

```python
plt.hist(x,bins=20,color="blue")
plt.axvline(x=_median,color="yellow")
```

```python
plt.hist(x,bins=20,color="blue")
plt.axvline(x=_mean,color="red")
plt.axvline(x=_median,color="yellow")
```

#mode

```python
_mode = stats.mode(x).mode
print("Mode :",_mode)
```

```python
plt.axvline(x=_mode)
```

```python
plt.hist(x,bins=20,color="blue")
plt.axvline(x=_mean,color="red")
plt.axvline(x=_median,color="yellow")
plt.axvline(x=_mode,color="green")
```
