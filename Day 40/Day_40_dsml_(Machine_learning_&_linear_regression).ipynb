```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib as mpl
from matplotlib import pyplot as plt
```

```python
data_df = pd.read_csv('https://www.statlearning.com/s/Advertising.csv')
```

```python
data_df
```

```python
fig = plt.figure(figsize=(15,4))
gs = mpl.gridspec.GridSpec(1, 3)    #row*column

#plot of sales vs tv
ax1 = fig.add_subplot(gs[0])
ax1.scatter(data_df['TV'], data_df['sales'],color="blue",marker=".")
ax1.set_xlabel('TV')
ax1.set_ylabel('sales')

#plot of sales vs radio
ax2 = fig.add_subplot(gs[1])
ax2.scatter(data_df['radio'], data_df['sales'],color="red",marker="*")
ax2.set_xlabel('radio')
ax2.set_ylabel('sales')

#plot of sales vs newspaper
ax3 = fig.add_subplot(gs[2])
ax3.scatter(data_df['newspaper'], data_df['sales'],color="green",marker="o")
ax3.set_xlabel('newspaper')
ax3.set_ylabel('sales')
```

```python
X = data_df.TV.to_numpy()
Y = data_df.sales.to_numpy()
```

```python
x_mean = X.mean()
y_mean = Y.mean()
```

```python
beta_1 = np.sum((X-x_mean)*(Y-y_mean))/np.sum((X-x_mean)**2)
beta_0 = y_mean - beta_1*x_mean
```

```python
print(f"y = {beta_0:.2f} + {beta_1:.2f}x")
```

```python
plt.scatter(X,Y,color="green")
plt.plot(X,beta_0+beta_1*X,color="red")
plt.show()
```
