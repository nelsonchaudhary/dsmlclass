#Operator Overloading

Addition(__ add __) operator

```python


class Point:
  def __init__(self,x,y):
    self.x = x
    self.y = y

  def __add__(self,other):
    x = self.x + other.x
    y = self.y + other.y
    return Point(x,y)

  def __str__(self):
    return f"({self.x},{self.y})"
```

```python
#create point object

p1 = Point(1,2)
p2 = Point(3,4)
```

```python
p3 = p1 + p2
print(p3)
```

```python
import operator

a = 10
b = 5
print(operator.add(a,b))
print(operator.__add__(a,b))
```

Subtraction(__ sub __)  operator

```python
class Point:
  def __init__(self,x,y):
    self.x = x
    self.y = y

  def __sub__(self,other):
    x = self.x - other.x
    y = self.y - other.y
    return Point(x,y)

  def __str__(self):
    return f"({self.x},{self.y})"

p1 = Point(9,8)
p2 = Point(5,6)
p3 = p1 - p2
print(p3)
```

Multiplication(__ mul __) operator

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __mul__(self, other):
        x = self.x * other.x
        y = self.y * other.y
        return Point(x, y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(9,8)
p2 = Point(5,6)
p3 = p1 * p2
print(p3)
```

Division(__ truediv __) operator

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __truediv__(self, other):
        x = self.x / other.x
        y = self.y / other.y
        return Point(x, y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(9,8)
p2 = Point(3,2)
p3 = p1 / p2
print(p3)
```

Floor Division(__ floordiv __) operator

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __floordiv__(self, other):
        x = self.x // other.x
        y = self.y // other.y
        return Point(x, y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(9,8)
p2 = Point(2,3)
p3 = p1 // p2
print(p3)
```

Modulo(__ mod __) operator

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __mod__(self, other):
        x = self.x % other.x
        y = self.y % other.y
        return Point(x, y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(9,8)
p2 = Point(5,3)
p3 = p1 % p2
print(p3)
```

Power( __ pow __ ) operator

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __pow__(self, other):
        x = self.x ** other.x
        y = self.y ** other.y
        return Point(x, y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(2,3)
p2 = Point(3,2)
p3 = p1 ** p2
print(p3)
```

Equal To( __ eq __) operator

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __eq__(self,other):
        return self.x == other.x and self.y == other.y

p1 = Point(9,8)
p2 = Point(9,8)

print(p1 == p2)
```

Less Than( __ lt __) operator

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __lt__(self,other):
        return self.x < other.x and self.y < other.y

p1 = Point(5,4)
p2 = Point(9,8)

print(p1 < p2)
```

Greater Than( __ gt __) operator

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __gt__(self,other):
        return self.x > other.x and self.y > other.y

p1 = Point(9,8)
p2 = Point(5,4)

print(p1 > p2)
```

Bitwise AND(__ and __) operator

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __and__(self,other):
        x = self.x & other.x
        y = self.y & other.y
        return Point(x,y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(6,5)
p2 = Point(3,1)

p3 = p1 & p2
print(p3)
```

Bitwise OR(__ or __) operator

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __or__(self,other):
        x = self.x | other.x
        y = self.y | other.y
        return Point(x,y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(6,5)
p2 = Point(3,1)

p3 = p1 | p2
print(p3)
```

Bitwise NOT(__ invert __) operator

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __invert__(self):
        x = ~self.x
        y = ~self.y
        return Point(x,y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(5,5)

p2 = ~p1
print(p2)
```

Bitwise XOR(__ xor __)

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __xor__(self,other):
        x = self.x ^ other.x
        y = self.y ^ other.y
        return Point(x,y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(6,5)
p2 = Point(3,1)

p3 = p1 ^ p2
print(p3)
```

Left Shift( __ lshift __) operator

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __lshift__(self,other):
        x = self.x << other.x
        y = self.y << other.y
        return Point(x,y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(2,3)
p2 = Point(2,1)

p3 = p1 << p2
print(p3)
```

Right Shift(__ rshift __) operator

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __rshift__(self,other):
        x = self.x >> other.x
        y = self.y >> other.y
        return Point(x,y)

    def __str__(self):
        return f"({self.x},{self.y})"

p1 = Point(8,6)
p2 = Point(2,1)

p3 = p1 >> p2
print(p3)
```
