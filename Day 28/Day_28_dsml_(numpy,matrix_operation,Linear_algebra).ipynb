```python
import numpy as np

matrix_1 = np.array([
    [1,2],
    [3,4]
])
matrix_1
```

```python
matrix_2 = np.array([
    [4,6],
    [7,8]
])
matrix_2
```

#element wise matrix addition

```python


result = matrix_1 + matrix_2
result
```

#element wise matrix subtraction


```python

matrix_sub = matrix_1 - matrix_2
matrix_sub
```

```python
# matrix_mult = matrix_1 * matrix_2
# matrix_mult                          wrong concept
```

#matrix multiplication


```python
result_mux = matrix_1 @ matrix_2
result_mux
```

```python
result_mux = np.matmul(matrix_1,matrix_2)
result_mux
```

```python
result_div = matrix_1 / matrix_2      #not recommended
result_div
```

```python

print(matrix_1)
print("///////////")
scalar = 2
result_add = matrix_1 + scalar
print(result_add)
```

#boardcasting


```python

matrix_3 = np.array([
    [1,2],
    [3,4]
])
array_1 = np.array([
    [5,6]
])

result_broad = matrix_3 + array_1
result_broad
```

#computing the determinant


```python

A = np.array([
    [1,2],
    [3,4]
])

det = np.linalg.det(A)
print(det)
```

#inverse
#we can compute the inverse of a matrix using the linalg.inv() function.

```python


a_inv = np.linalg.inv(A)
print(a_inv)
```

```python
print(A@a_inv)
```

```python
#qn

a = np.array([
    [2,1],
    [1,3]
])

b = np.array([5,6])
X = np.linalg.solve(a,b)
X
```

```python
#class task

# x+y+z=6
# 2x+5y+z=-4
# 2x+3y+8z=5


a = np.array([
    [1,1,1],
    [2,5,1],
    [2,3,8]
])

b = np.array([6,-4,5])
X = np.linalg.solve(a,b)
X
```

```python
a=np.array([
    [1,1],
    [2,1],
    [3,1]
])
b=np.array([2,3,5])

beta=np.linalg.inv(a.T@a)@a.T@b
beta
```

```python
print("slope(m)",beta[0])
print("intercept(c)",beta[1])
```
