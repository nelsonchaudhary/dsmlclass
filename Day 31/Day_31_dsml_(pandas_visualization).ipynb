using kind="bar"

```python
titanic_df["Survived"].value_counts().plot(
    kind="bar",
    title="count of survived passengers"
)
```

```python
titanic_df["Survived"].value_counts().plot(
    kind="bar",
    title="Count of Survived Passengers",
    figsize=(5, 5),
    yticks=list(range(0, 600, 50)),
    xlabel="Survived",
    ylabel="Count",
    grid=True,
    color="red"
)
```

```python
pd.crosstab(titanic_df["Survived"],titanic_df["Sex"]).plot(
    kind="bar",
    title="Count of Survived Passengers",
    figsize=(5, 5),
    legend=True,
    yticks=list(range(0, 400, 100)),
    xlabel="Sex",
    ylabel="No. of Passengers",
    grid=True,
    color=["grey", "green"]
)
```

PieChart

```python
titanic_df["Embarked"].value_counts().plot(
    kind="pie",
    title="Passengers by Embarked",
    autopct="%1.1f%%",
    figsize=(5, 5)
)
```

```python
ct = pd.crosstab(titanic_df["Survived"], titanic_df["Sex"])

ct.plot(
    kind="pie",
    subplots=True,
    title="Survival by Sex",
    figsize=(10, 5),
    autopct="%1.1f%%",

)
```
