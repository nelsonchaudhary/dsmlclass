```python
def condition(n):
  if n==0 or n==1:
    return 1
  return n*condition(n-1)
condition(5)
```

```python
def fibonacci(n, data={}):
    if n in data:
        return data[n]
    if n <= 1:
        return n
    data[n] = fibonacci(n-1, data) + fibonacci(n-2, data)
    return data[n]

for i in range(10):
    print(fibonacci(i), end=" ")
```

```python
#decorator

def decorator(func):
  def wrapper():
    print("Before")
    func()
    print("after")
  return wrapper

@decorator
def func():
  print("hello")
```

```python
func()
```

```python
#write a decorator login_required that checks whether a user is logged in. if the variable logged_in is True,execute the function:otherwise,
#print"access Denied".

def login_required(func):
  def wrapper():
    logged_in = True
    if logged_in:
      func()
    else:
      print("access denied")
  return wrapper

@login_required
def func():
  print("Welcome to the sensitive data area!")
```

```python
func()
```

```python
class A:
  def display(self):
    print("hello from base class")
```

```python
class B(A):
  def display1(self):
    print("hello i am from derived class")
```

```python
obj = B()
obj.display()
obj.display1()
```

```python
#hybrid inheritance

class A:
  def display0(self):
    print("A")
class B(A):
  def display1(self):
    print("B")
class C(A):
  def display2(self):
    print("C")
class D(B,C):
  def display3(self):
    super().display0()
    super().display1()
    super().display2()
    print("D")
```

```python
obj = D()
obj.display3()
```
