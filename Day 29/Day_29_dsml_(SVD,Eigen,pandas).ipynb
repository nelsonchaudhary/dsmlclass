#Eigen value & Eigen vecotr

```python
import numpy as np
```

```python
A = np.array([
    [1,2],
    [3,4]
])

eigenvalues, egienvectors = np.linalg.eig(A)

print(eigenvalues)
print(egienvectors)
```

#singular value Decomposition(SVD)<br>
used in data compression

```python
U ,S ,VT = np.linalg.svd(A)
print("U matrix:  \n",U)
print("Sigma matrix:  \n",S)
print("VT matrix:  \n",VT)
```

```python
print(U @ U.T)
```

## Difference between `loc` and `iloc`

In pandas, both `.loc` and `.iloc` are used for selecting data from a DataFrame, but they differ in how they interpret their indexing arguments:

*   **`.loc` (Label-based indexing)**:
    *   Selects data based on the **labels** of rows and columns.
    *   Can accept a single label, a list of labels, a slice of labels, or a boolean array.
    *   The end of the slice is **inclusive** when using labels (e.g., `df.loc['A':'C']` includes 'C').

*   **`.iloc` (Integer-location based indexing)**:
    *   Selects data based on the **integer positions** of rows and columns.
    *   Can accept a single integer, a list of integers, a slice of integers, or a boolean array.
    *   The end of the slice is **exclusive** when using integers (e.g., `df.iloc[0:3]` includes rows 0, 1, 2 but not 3), similar to Python's standard slicing.

```python
# Example of iloc: Select the first two rows and all columns
res_iloc = df.iloc[0:2]
print("DataFrame selected using iloc (first two rows):\n", res_iloc)

# Example of iloc: Select row at index 1 and column at index 0
# (which corresponds to 'ram' and 'name' respectively)
value_iloc = df.iloc[1, 0]
print("\nValue selected using iloc (row index 1, column index 0):", value_iloc)
```
