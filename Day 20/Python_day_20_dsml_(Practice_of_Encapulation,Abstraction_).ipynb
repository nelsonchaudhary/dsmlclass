```python
class BankAccount:
    def __init__(self, name, account, pin, balance):
        self.name = name
        self._account_type = account
        self.__pin = pin
        self.__balance = balance

    def verify_pin(self, pin):
        if self.__pin == pin:
            return True
        else:
            return False

    def get_account(self):
        return self._account_type

    def get_balance(self):
        return self.__balance


class SavingAccount(BankAccount):
    def __init__(self, name, account, pin, balance, interest):
        super().__init__(name, account, pin, balance)
        self.interest = interest

    def calculate_interest(self):
        return self.get_balance() * self.interest

    def add_interest(self):
      total_amount = self.get_balance() + self.calculate_interest()
      return total_amount


# Object
obj = SavingAccount("nelson", "Saving", 2222, 2000, 0.2)

# Ask user for PIN
user_pin = int(input("Enter your PIN: "))

if obj.verify_pin(user_pin):     #obj.veriy_pin(user_pin) it checks wheather the pin matches or not
    print("Access Granted")
    print("Account Type:", obj.get_account())
    print("Balance:", obj.get_balance())
    print("Interest:", obj.calculate_interest())
    print("amount after interest:", obj.add_interest())
else:
    print("Incorrect PIN. Access Denied.")
```

#Class task
Create an abstract BankAccount class with abstract methods deposit() and withdraw().Implement them in a SavingAccount class.

```python
#class task
from abc import ABC, abstractmethod

class BankAccount(ABC):

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass


class SavingAccount(BankAccount):

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Current Balance:", self.balance)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Current Balance:", self.balance)
        else:
            print("Insufficient Balance")
```

```python
obj = SavingAccount(5000)

obj.deposit(1000)
obj.withdraw(2000)
obj.withdraw(5000)
```
