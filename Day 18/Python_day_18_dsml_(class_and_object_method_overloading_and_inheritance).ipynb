```python
#class task
#multiple inheritance

class Hot_drinks:
  def hot(self):
    print("keeps you warm")

class Cold_drinks:
  def cold(self):
    print("keeps you cool")

class Drinks(Hot_drinks,Cold_drinks):
  def drinkable(self):
    print("helps for hydration")


a=Drinks()
a.hot()
a.cold()
a.drinkable()
```

```python
# multilevel inheritance

class Dark_chocolate:
  def dark(self):
    print("have the taste of coco")
class Mango_Chocolate(Dark_chocolate):
  def mango(self):
    print("have the taste of mango")
class Kiwi_chocolate(Mango_Chocolate):
  def kiwi(self):
    print("have the taste of kiwi")

a=kiwi_chocolate()
a.dark()
a.mango()
a.kiwi()
```

```python
#hierarchical inheritance
class Drinks:
  def drinkable(self):
    print("helps for hydration")
class Hot_drinks(Drinks):
  def hot(self):
    print("keeps you warm")
class Cold_drinks(Drinks):
  def cold(self):
    print("keeps you cool")

a=Hot_drinks()
a.drinkable()
a.hot()
b=Cold_drinks()
b.drinkable()
b.cold()

```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print(f"Name: {self.name}, Age: {self.age}")


class Teacher(Person):
    def __init__(self, name, age, subject):
        Person.__init__(self, name, age)
        self.subject = subject

    def show(self):
        Person.show(self)
        print(f"Subject: {self.subject}")


class Student(Person):
    def __init__(self, name, age, student_id):
        Person.__init__(self, name, age)
        self.student_id = student_id

    def show(self):
        Person.show(self)
        print(f"Student ID: {self.student_id}")


class TeachingAssistance(Teacher, Student):
    def __init__(self, name, age, subject, student_id, department, faculty):
        super().__init__(name, age, subject)   # calls Teacher.__init__ via MRO
        self.student_id = student_id
        self.department = department
        self.faculty = faculty

    def show(self):
        super().show()
        print(f"Student ID: {self.student_id}")
        print(f"Department: {self.department}")
        print(f"Faculty: {self.faculty}")


TA = TeachingAssistance("Tony", 22, "Python", 101, "dsml", "Engineering")
TA.show()
```

#1.Create a class Person with attributes name and age. Create a class Student that inherits from Person and adds a grade attribute.Display all details.


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print(f"Name: {self.name} , Age: {self.age}")


class Student(Person):
    def __init__(self, name, age, grade):
        super().__init__(name, age)
        self.grade = grade

    def show(self):
        super().show()
        print(f"Grade: {self.grade}")
```

```python
obj = Student("Nelson", 22, "10")
obj.show()
```

#4.Create a class Shape with a method display(). Create two child classes Circle and Rectangle that inherit from Shape and implement their own area calculation methods

```python


class Shape:
  def display(self):
    print("this is a shape")

class Circle(Shape):
  def __init__(self, radius):
    self.radius = radius

  def area(self):
    result = 3.14 * self.radius ** 2
    print(f"area of circle = {result}")

class Rectangle(Shape):
  def __init__(self, length, width):
    self.length = length
    self.width = width

  def area(self):
    result = self.length * self.width
    print(f"area of rectangle = {result}")


c = Circle(5)
c.display()
c.area()

r = Rectangle(4,6)
r.display()
r.area()
```
