#Class task-1 : Calculator Using Different Types of Attributes and Methods


Create a Python class named Calculator that demonstrates the use of different types of attributes and methods.<br>
Requirements:<br>
1.Create a class attribute named company and assign it any company name.<br>
2.Use the __init __() method to:
Ask the user to enter two numbers.<br>
Ask the user to enter an operator (+,-,*, /).<br>
3.Create an instance method named calculate() that performs the appropriate operation based on the operator entered by the user.<br>
4.Create a class method named show_company() that displays the company name.<br>
5.Create a static method named welcome() that displays a welcome message.<br>
6.Create an object of the class and call all the required methods.<br>

```python
class Calculator:
    company = "Google"

    @classmethod
    def company_name(cls):
        print("Company:", cls.company)

    @staticmethod
    def welcome():
        print("Welcome to Calculator")

    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def calculate(self):
        print("Operators:")
        print("+ -> Addition")
        print("- -> Subtraction")
        print("* -> Multiplication")
        print("/ -> Division")

        choice = input("Enter an operator: ")

        if choice == "+":
            add = self.num1 + self.num2
            print("The sum is", add)

        elif choice == "-":
            subtract = self.num1 - self.num2
            print("The subtracted value is", subtract)

        elif choice == "*":
            multiply = self.num1 * self.num2
            print("The multiplied value is", multiply)

        elif choice == "/":
          divide = self.num1 / self.num2
          print("the quoitent is ",divide)
        else:
            print("Invalid operator")
```

```python
obj = Calculator(10,5)

obj.welcome()
Calculator.company_name()
obj.calculate()
```

#Class task-2 : Student Grade Management System
Create a Python class named Student.<br>
Requirements:<br>
1.Create a class attribute named school.<br>
2.Use the __init__() method to take:<br>
Student name<br>
Student marks<br>
3.Create an instance method named check_grade() that displays:<br>
A Grade: Marks ≥ 80<br>
B Grade: Marks ≥ 60 and < 80<br>
C Grade: Marks ≥ 40 and < 60<br>
Fail: Marks < 40<br>
4.Create a class method to display the school name.<br>
5.Create a static method to display a welcome message.<br>
6.Create an object and display the student's grade.<br>



```python
class Student:
  school = "skillshikshya"

  def __init__(self,name,marks):
    self.name = name
    self.marks = marks

  def check_grade(self):
    if 80 <= self.marks <= 100:
      print("Student Name :", self.name)
      print("Marks :", self.marks)
      print("Grade Obtained : A")
    elif 60 <= self.marks < 80:
      print("Student Name :", self.name)
      print("Marks :", self.marks)
      print("Grade Obtained : B")
    elif 40 <= self.marks < 60:
      print("Student Name :", self.name)
      print("Marks :", self.marks)
      print("Grade Obtained : C")
    else:
      print("Student Name :", self.name)
      print("Marks :", self.marks)
      print("Fail")


  @classmethod
  def school_name(cls):
    print("School : ",cls.school)

  @staticmethod
  def welcome():
    print("welcome to skillshikshya")


obj =Student("Nelson",99)
obj1 =Student("Ram",77)

obj.welcome()
Student.school_name()
obj.check_grade()
obj1.check_grade()

```

Class task-3 : Bank Account Management System
Create a Python class named BankAccount.<br>
Requirements:<br>
1.Create a class attribute named bank_name.<br>
2.Use the __init__() method to take:<br>
Account holder name<br>
Initial balance<br>
3.Create the following instance methods:<br>
deposit() → Ask the user for an amount and add it to the balance.<br>
withdraw() → Ask the user for an amount and deduct it from the balance if sufficient balance is available.<br>
show_balance() → Display the current balance.<br>
4.Create a class method to display the bank name.<br>
5.Create a static method to display banking rules or a welcome message.<br>
6.Create an object and test all methods.<br>
<br>
<br>
Sample output:<br>
Welcome to RAD Bank<br>
Enter account holder name: Dhiraj<br>
Enter initial balance: 10000<br>
Enter deposit amount: 5000<br>
Current Balance: 15000<br>
Enter withdrawal amount: 3000<br>
Current Balance: 12000


```python
class BankAccount:
    bank_name = "Global IME Limited"

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def get_service(self):
        print("====services====")
        print("1.deposit")
        print("2.withdraw")
        print("3.Show_balance")
        print("4.Exit")

        while True:
            choice = int(input("enter the choice of service = "))

            if choice == 1:
                deposit_amount = float(input("enter the amount you want to deposit = "))
                self.balance += deposit_amount
                print("your new balance is ", self.balance)
            elif choice == 2:
                withdraw_amount = float(input("enter the amount you want to withdraw = "))
                if self.balance >= withdraw_amount:
                    self.balance -= withdraw_amount
                    print("your new balance is ", self.balance)
                else:
                    print("Insufficient balance")
            elif choice == 3:
                print("your current balance is ", self.balance)
            elif choice == 4:
                print("Thank You")
                break
            else:
                print("invalid choice")

    @classmethod
    def bank(cls):
        print("Bank Name = ", cls.bank_name)

    @staticmethod
    def greet():
        print("Welcome to Global IME Limited Bank")
```

```python
obj = BankAccount("Nelson",50000)
obj.greet()
obj.bank()
obj.get_service()
```
