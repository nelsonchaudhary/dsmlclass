```python
import matplotlib.pyplot as plt
```

```python
min_length = iris_df['SepalLengthCm'].min()
max_length = iris_df['SepalLengthCm'].max()
```

```python
normal_length = (iris_df.SepalLengthCm-min_length)/(max_length-min_length)
```

```python
normal_length
```

```python
normal_length.plot.kde()
```

```python
iris_df["SepalLengthCm"].plot.kde()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
min_max_scaler = MinMaxScaler(feature_range=(0,1))
```

```python
norm_data = min_max_scaler.fit_transform(iris_df.drop("Species",axis=1))
```

```python
norm_data
```

```python
#drawback of normalization
# if agges are:
# 20,30,44,1000   #1000 is a outlier

#highly sensitive  to outlier (Biggest Drawback)
#no robust for real world data

# advantage
#propertional spacing

#Monotonicity is a mathematical property describing a function or process that moves in one consistent direction—either
# entirely going up or staying flat, or entirely going down or staying flat, without switching back and forth.
```

```python
pd.DataFrame(norm_data,columns=["Id","SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"])
```

```python
pd.DataFrame(norm_data,columns=["Id","SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]).describe()
```

```python
# Z score Normilazition

mean = iris_df.SepalLengthCm.mean()
std = iris_df.SepalLengthCm.std()

Z_norm = (iris_df.SepalLengthCm-mean)/std
Z_norm
```

```python
round(Z_norm.describe(),2)
```

```python
Z_norm.plot.kde()
```
