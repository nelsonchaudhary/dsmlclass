#Method overriding in python

```python
class Parent:
  def myMethods(self):
    print("Calling parent method")
```

```python
class Child(Parent):
  def myMethods(self):
    super().myMethods()
    print("Calling from child methods")
```

```python
c = Child()
c.myMethods()
```

```python
class Employee:
  def __init__(self,nm,sal):
    self.name=nm
    self.salary=sal

  def getName(self):       #Camel Case
    return self.name
  def getSalary(self):
    return self.salary
```

```python
class Salesoffice(Employee):
  def __init__(self,nm,sal,inc):
    super().__init__(nm,sal)
    self.incnt = inc
  def getSalary(self):
    return self.salary + self.incnt
```

```python
e1 = Employee("Nelson",50000)
print("total salary of {} is RS {}".format(e1.getName(),e1.getSalary()))
```

```python
s1 = Salesoffice("ram",600000,100000)
print("total salary of {} is RS {}".format(s1.getName(),s1.getSalary()))
```

```python
class Example:
  def __init__(self):
    self.public_var = 1
    self.__private_var = 2
    self._protected_var = 3

  def public_method(self):
    print("public method")
    print("private variable",self.__private_var)

  def __private_method(self):
    print("private method")

  def _protected_method(self):
    print("protected method")
```

```python
obj = Example()
```

```python
#accessing public members
print(obj.public_var)
obj.public_method()
```

```python
#accessing protected members (possible not recommended)
print(obj._protected_var)
obj._protected_method()
```

```python
#accessing private member directly
print(obj.__private_var)
obj.__private_method()                        # we cannot directly access private method
```

```python
#accessing private member using name mangling
print(obj._Example__private_var)
obj._Example__private_method()
```

```python
class BankAccount:
  def __init__(self,name,balance):
    self.name = name
    self.__initial_balance = balance

  def get_name(self):
    print("Name = ",self.name)
  def get_balance(self):
    return self.__initial_balance

  def get_service(self):
        print("====services====")
        print("1.deposit")
        print("2.withdraw")
        print("3.Check_balance")
        print("4.Exit")

        while True:
            choice = int(input("enter the choice of service = "))

            if choice == 1:
                deposit_amount = float(input("enter the amount you want to deposit = "))
                self.__initial_balance += deposit_amount
                print("your new balance is ", self.__initial_balance)
            elif choice == 2:
                withdraw_amount = float(input("enter the amount you want to withdraw = "))
                if self.__initial_balance >= withdraw_amount:
                    self.__initial_balance -= withdraw_amount
                    print("your new balance is ", self.__initial_balance)
                else:
                    print("Insufficient balance")
            elif choice == 3:
                print("your current balance is ", self.__initial_balance)
            elif choice == 4:
                print("Thank You")
                break
            else:
                print("invalid choice")
```

```python

class SavingAccount(BankAccount):
  def __init__(self,name,balance,interest_rate):
    super().__init__(name,balance)
    self.interest_rate = interest_rate

  def calculate_rate(self):
    amount = self.get_balance() * self.interest_rate
    return amount
```

#Abstraction

```python
#abstraction

from abc import ABC,abstractmethod

class Animal(ABC):
  @abstractmethod
  def make_sound(self):
    pass

class Dog(Animal):
  def make_sound(self):
    print("woof")

class Cat(Animal):
  def make_sound(self):
    print("meow")
```

```python
d = Dog()
d.make_sound()
c=Cat()
c.make_sound()
```
