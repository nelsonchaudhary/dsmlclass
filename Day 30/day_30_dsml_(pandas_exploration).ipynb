### 1. Survived Passengers by Gender

```python
survived_df = titanic_df[titanic_df['Survived'] == 1].copy()
```

```python
print(survived_df['Sex'].value_counts())
```

### 2. Survived Passengers by Pclass

```python
print(survived_df['Pclass'].value_counts())
```

### 3. Survived Passengers by Age Group

```python
# Create age groups
age_bins = [0, 12, 18, 60, titanic_df['Age'].max()]
age_labels = ['Child', 'Teenager', 'Adult', 'Senior']
survived_df['AgeGroup'] = pd.cut(survived_df['Age'], bins=age_bins, labels=age_labels, right=False)

print(survived_df['AgeGroup'].value_counts())
```

### 4. Survived Passengers by Fare Group

```python
fare_bins = [0, 10, 30, 100]
fare_labels = ['very low', 'low', 'high']

survived_df['FareGroup'] = pd.cut(
    survived_df['Fare'],
    bins=fare_bins,
    labels=fare_labels,
    right=False
)

print(survived_df['FareGroup'].value_counts())
```
